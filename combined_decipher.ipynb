```python
message=""
reverse=""
key_known=""
alphabet="abcdefghijklmnopqrstuvwxyz"
key=0
get_letter=0
original=[]

message=str(input("Please enter the encrypted message: "))
i=len(message)-1

while i >= 0:
  reverse=reverse+message[i]
  i=i-1

key_known=str(input("Was the key known? Y/N: ")).strip().upper()

if key_known == "Y":
  key=int(input("Enter the key used: "))
  if 0 < key <= 26:
    for letter in reverse:
      get_letter=(alphabet.index(letter)-key) % 26
      original.append(alphabet[get_letter])
    print("".join(original) + " is the original message")
  else:
    print("Incorrect. Key can only be between 1 and 26")
elif key_known == "N":
  for key in range (26):
    original=[]
    for letter in reverse:
      get_letter=(alphabet.index(letter)-key) % 26
      original.append(alphabet[get_letter])
    print("")
    print("".join(original))
  print("One of these will be the original message.")
else:
    print("Incorrect. Only Y or N can be processed")
```
