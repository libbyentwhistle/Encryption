```python
message = "The colour purple is the best"
#message to be translated

translated = ""
#declaring translated variable

i=len(message) -1
#this calculates the number of characters in the variable message and subtracts 1

while i>=0: #whilst the value of i is greater than or equal to 0
  translated=translated + message[i]
  i= i - 1
print(translated)
```
