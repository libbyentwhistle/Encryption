```python
alphabet = "abcdefghijklmnopqrstuvwxyz"
get_letter=0
translated = []

encrypted = str(input("Enter the encrypted message here: ")).strip().lower()
print("")
key=int(input("Enter the key that was used to encrypt the message: "))

if 0 < key <= 26:
  for letter in encrypted:
    get_letter=(alphabet.index(letter)-key)% 26
    translated.append(alphabet[get_letter])
  print("".join(translated) + " is the decrypted message")
else:
  print("Incorrect. Key must be between 1 and 26")
```
