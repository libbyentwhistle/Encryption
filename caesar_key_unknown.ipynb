```python
message=""
alphabet="abcdefghijklmnopqrstuvwxyz"
get_letter=0
translated=[]
key_known=""
key=0

message=str(input("Please enter the encrypted message: "))
key_known=str(input("Do you know what key was used to encrypt this message (Y/N): ")).strip().upper()

if key_known == "Y":
  key=int(input("Enter the key used to encrypt this message: "))
  if 0 < key <= 26:
      for letter in message:
        get_letter=(alphabet.index(letter)-key) % 26
        translated.append(alphabet[get_letter])
      print("".join(translated)+ " is the decrypted message")
  else:
      print("Incorrect. The key can only be between 1 and 26")
elif key_known == "N":
  for key in range (26):
    translated=[]
    for letter in message:
      get_letter=(alphabet.index(letter)-key) % 26
      translated.append(alphabet[get_letter])
    print("".join(translated))
    print("")
  print("One of these will be the original message")
else:
  print("Incorrect. Only inputs Y and N can be processed")
```
