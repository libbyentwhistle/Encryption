```python
alphabet = "abcdefghijklmnopqrstuvwxyz"
get_letter = 0
keyword=[]
#declaring variables

text=str(input("Enter the message you wish to be encrypted here: ")).strip().lower()
print("")
key=int(input("Enter the key you wish to use to encrypt the message here: "))
#asking for inputs

if 0 < key <= 26:
  for letter in text:
    get_letter= (alphabet.index(letter) + key) % 26
    keyword.append(alphabet[get_letter])
else:
    print("Incorrect. Key must be between 1 and 26")
#ensuring that an incorrect value for key cannot be entered.

translated=""
i=len(keyword)-1
#setting up for the reverse cipher

while i>=0:
  translated=translated + keyword[i]
  i=i-1
print(translated)
#reversing the caesar shift and printing it out
```
